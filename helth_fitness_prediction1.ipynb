```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from sklearn.linear_model import  LinearRegression
from sklearn.model_selection import  train_test_split
from sklearn.metrics import r2_score
```

```python
df=pd.read_csv("health_fitness_dataset.csv")
df
```

```python
df.isnull().sum()
```

```python
# convert all columns to numeric
df = df.apply(pd.to_numeric, errors="coerce")

# remove negative exercise minutes
df = df[df["exercise_minutes"] >= 0]

# fill missing values
df["exercise_minutes"] = df["exercise_minutes"].fillna(df["exercise_minutes"].mean())

# Quartiles
Q1 = df["exercise_minutes"].quantile(0.25)
Q3 = df["exercise_minutes"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

# find outliers
outlier = df[(df["exercise_minutes"] < lower) |(df["exercise_minutes"] > upper)]
print(outlier)
```

```python
x=df[["exercise_minutes","steps","food_calories","sleep_hours","water_intake_liters"]]
y=df["weight"]

print(x.head())
print(y.head())
```

```python
x_train,x_test,y_train,y_test=train_test_split(
    x,y,test_size=0.2,random_state=42
)
```

```python
model=LinearRegression()
```

```python
model.fit(x_train,y_train)  
```

```python
new_data = [[45, 8000, 2200, 7, 2.5]]
prediction = model.predict(new_data)
print(prediction[0])
```

```python
v=model.predict(x_test)
score=r2_score(y_test,v)
print(score)
```

```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv("Bangalore.csv")
df
```

```python
print("head\n",df.head()) 
print("shape\n",df.shape) 
print("info\n",df.info()) 
print("des\n",df.describe())
```

```python
print(df.isnull().sum())
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
df
```

```python
df.rename(columns={
    'No. of Bedrooms': 'Bedrooms'
}, inplace=True)
```

```python
df.columns
```

```python
numeric_columns = df.select_dtypes(include=np.number).columns

print(numeric_columns)


for col in numeric_columns:

    q1 = df[col].quantile(0.25)

    q3 = df[col].quantile(0.75)

    iqr = q3 - q1

    lower = q1 - 1.5 * iqr

    upper = q3 + 1.5 * iqr

    df = df[
        (df[col] >= lower) &
        (df[col] <= upper)
    ]


if 'Price' in df.columns:
    df = df[df['Price'] > 0]

if 'Area' in df.columns:
    df = df[df['Area'] > 0]

print(df.shape)
```

```python
df.to_csv("clean_Bangalore.csv", index=False)
```
